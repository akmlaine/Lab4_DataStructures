```python

# APPENDIX A: ENGINEERING TEST DATA ORGANIZER

LAST_NAME = "Malitao"
FAVORITE_ARTIST = "Adie"
SEED_NUM = 5

sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

records = []

# 1. Generate test data
for i, source in enumerate(sources):
    for j, category in enumerate(categories):
        letter = FAVORITE_ARTIST[(i + j) % len(FAVORITE_ARTIST)]
        value = SEED_NUM * 10 + ord(letter) % 50

        records.append([source, category, value])

        if category in ["TEMPERATURE", "VOLTAGE"]:
            records.append([source, category, value + SEED_NUM])

# Add one repeated record
records.append(records[0].copy())

# 2. Display generated data
print("GENERATED TEST DATA")
print(f"{'SOURCE':<18}{'CATEGORY':<15}{'VALUE':>8}")
for source, category, value in records:
    print(f"{source:<18}{category:<15}{value:>8}")

# 3. Find repeated and distinct records
seen = set()
repeated = []

for record in records:
    item = tuple(record)
    if item in seen:
        repeated.append(item)
    seen.add(item)

print("\nREPEATED RECORDS")
for item in set(repeated):
    print(item)

print("\nTOTAL DISTINCT RECORDS:", len(seen))

# 4. Numerical summary
values = [record[2] for record in records]

print("\nNUMERICAL SUMMARY")
print("Total measurements:", len(records))
print("Minimum:", min(values))
print("Maximum:", max(values))
print("Average:", round(sum(values) / len(values), 2))

# 5. Organize and summarize by category
print("\nCATEGORY SUMMARY")
print(f"{'CATEGORY':<15}{'COUNT':>8}{'AVERAGE':>12}")

for category in categories:
    category_values = [
        r[2] for r in records if r[1] == category
    ]
    average = sum(category_values) / len(category_values)

    print(f"{category:<15}{len(category_values):>8}{average:>12.2f}")

# 6. Order results
ordered = sorted(records, key=lambda r: (r[0], r[1], r[2]))

print("\nORDERED RESULTS")
for source, category, value in ordered:
    print(f"{source:<18}{category:<15}{value:>8}")

# 7. Final summary
print("\nFINAL SUMMARY")
print("Number of sources:", len(sources))
print("Number of categories:", len(categories))
print("Total records:", len(records))
print("Distinct records:", len(seen))
print("Repeated records:", len(repeated))
print("Overall average:", round(sum(values) / len(values), 2))
```

```python
# APPENDIX 2: STUDENT PERFORMANCE DATA ANALYZER

LAST_NAME = "Malitao"
FAVORITE_ARTIST = "Adie"
SEED_NUM = 5

students = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]

subjects = [
    "PROGRAMMING",
    "MATHEMATICS",
    "ELECTRONICS"
]

# Generate student performance records
records = []

for i, student in enumerate(students):
    for j, subject in enumerate(subjects):
        artist_char = FAVORITE_ARTIST[(i + j) % len(FAVORITE_ARTIST)]

        score = 60 + (
            ord(artist_char) + SEED_NUM + i * 17 + j * 13
        ) % 41

        records.append({
            "student": student,
            "subject": subject,
            "score": score
        })

# Add one missing record
records[0]["score"] = None

# Add one repeated record
records.append(records[1].copy())


# Classify student performance
def get_category(score):
    if score is None:
        return "Incomplete"
    elif score >= 90:
        return "Excellent"
    elif score >= 80:
        return "Very Good"
    elif score >= 75:
        return "Passed"
    else:
        return "Failed"


# 1. Display generated performance records
print("1. GENERATED PERFORMANCE RECORDS")

for record in records:
    print(record)


# 2. Organize data by student and subject
print("\n2. ORGANIZED STUDENT/SUBJECT DATA")

organized_data = {}

for record in records:
    student = record["student"]
    subject = record["subject"]

    if student not in organized_data:
        organized_data[student] = {}

    if subject not in organized_data[student]:
        organized_data[student][subject] = []

    organized_data[student][subject].append(record["score"])

for student, subject_data in organized_data.items():
    print(student, subject_data)


# 3. Identify repeated and incomplete records
print("\n3. REPEATED AND INCOMPLETE RECORDS")

seen = set()
repeated_records = []
incomplete_records = []

for record in records:
    key = (
        record["student"],
        record["subject"],
        record["score"]
    )

    if key in seen:
        repeated_records.append(record)
    else:
        seen.add(key)

    if record["score"] is None:
        incomplete_records.append(record)

print("Repeated records:", repeated_records)
print("Incomplete records:", incomplete_records)


# 4. Summarize performance categories
print("\n4. PERFORMANCE CATEGORIES")

category_counts = {
    "Excellent": 0,
    "Very Good": 0,
    "Passed": 0,
    "Failed": 0,
    "Incomplete": 0
}

for record in records:
    category = get_category(record["score"])
    category_counts[category] += 1

print(category_counts)


# 5. Student summaries
print("\n5. STUDENT SUMMARIES")

for student in students:
    student_records = [
        record for record in records
        if record["student"] == student
    ]

    valid_scores = [
        record["score"] for record in student_records
        if record["score"] is not None
    ]

    average = (
        sum(valid_scores) / len(valid_scores)
        if valid_scores else 0
    )

    print(
        student,
        "| Valid scores:", len(valid_scores),
        "| Average:", round(average, 2)
    )


# 6. Subject summaries
print("\n6. SUBJECT SUMMARIES")

for subject in subjects:
    subject_records = [
        record for record in records
        if record["subject"] == subject
    ]

    valid_scores = [
        record["score"] for record in subject_records
        if record["score"] is not None
    ]

    average = (
        sum(valid_scores) / len(valid_scores)
        if valid_scores else 0
    )

    print(
        subject,
        "| Valid scores:", len(valid_scores),
        "| Average:", round(average, 2)
    )


# 7. Ordered results
print("\n7. ORDERED RESULTS")

ordered_records = sorted(
    records,
    key=lambda record: (
        record["score"] is None,
        -(record["score"] or 0),
        record["student"],
        record["subject"]
    )
)

for record in ordered_records:
    print(
        record["student"],
        "|", record["subject"],
        "| Score:", record["score"],
        "| Category:", get_category(record["score"])
    )


# 8. Final summary
print("\n8. FINAL SUMMARY")

valid_scores = [
    record["score"] for record in records
    if record["score"] is not None
]

print("Total records:", len(records))
print("Unique records:", len(seen))
print("Repeated records:", len(repeated_records))
print("Incomplete records:", len(incomplete_records))
print("Highest score:", max(valid_scores) if valid_scores else "N/A")
print("Lowest score:", min(valid_scores) if valid_scores else "N/A")
print("Overall average:", round(sum(valid_scores) / len(valid_scores), 2)
      if valid_scores else "N/A")
print("Performance categories:", category_counts)
```

```python
# APPENDIX 3: TRANSACTION HISTORY ANALYZER

LAST_NAME = "Malitao"
FAVORITE_ARTIST = "Adie"
SEED_NUM = 5

customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]

products = {
    "NOTEBOOK": 50,
    "PEN": 15,
    "USB": 250,
    "HEADPHONES": 500,
    "CALCULATOR": 350
}

product_names = list(products.keys())

# Generate transaction records
transactions = []

for i in range(10):
    artist_char = FAVORITE_ARTIST[i % len(FAVORITE_ARTIST)]

    customer = customers[i % len(customers)]

    product_index = (
        ord(artist_char) + SEED_NUM + i
    ) % len(product_names)

    product = product_names[product_index]

    quantity = (
        ord(artist_char) + SEED_NUM + i * 3
    ) % 5 + 1

    total = products[product] * quantity

    transactions.append({
        "transaction_id": f"T{i + 1}",
        "customer": customer,
        "product": product,
        "quantity": quantity,
        "total": total
    })

# Add one missing transaction record
transactions[0]["quantity"] = None
transactions[0]["total"] = None

# Add one repeated transaction
transactions.append(transactions[1].copy())


# Classify transaction amounts
def get_transaction_category(total):
    if total is None:
        return "Incomplete"
    elif total < 100:
        return "Low"
    elif total < 500:
        return "Medium"
    else:
        return "High"


# 1. Display generated transaction records
print("1. GENERATED TRANSACTION RECORDS")

for transaction in transactions:
    print(transaction)


# 2. Organize transactions by customer and product
print("\n2. ORGANIZED CUSTOMER/PRODUCT DATA")

organized_data = {}

for transaction in transactions:
    customer = transaction["customer"]
    product = transaction["product"]

    if customer not in organized_data:
        organized_data[customer] = {}

    if product not in organized_data[customer]:
        organized_data[customer][product] = []

    organized_data[customer][product].append(transaction["total"])

for customer, product_data in organized_data.items():
    print(customer, product_data)


# 3. Identify repeated and incomplete transactions
print("\n3. REPEATED AND INCOMPLETE TRANSACTIONS")

seen = set()
repeated_transactions = []
incomplete_transactions = []

for transaction in transactions:
    key = (
        transaction["transaction_id"],
        transaction["customer"],
        transaction["product"],
        transaction["quantity"],
        transaction["total"]
    )

    if key in seen:
        repeated_transactions.append(transaction)
    else:
        seen.add(key)

    if (
        transaction["quantity"] is None
        or transaction["total"] is None
    ):
        incomplete_transactions.append(transaction)

print("Repeated transactions:", repeated_transactions)
print("Incomplete transactions:", incomplete_transactions)


# 4. Categorize transaction amounts
print("\n4. TRANSACTION CATEGORIES")

category_counts = {
    "Low": 0,
    "Medium": 0,
    "High": 0,
    "Incomplete": 0
}

for transaction in transactions:
    category = get_transaction_category(transaction["total"])
    category_counts[category] += 1

print(category_counts)


# 5. Customer summaries
print("\n5. CUSTOMER SUMMARIES")

for customer in customers:
    customer_transactions = [
        transaction for transaction in transactions
        if transaction["customer"] == customer
    ]

    valid_totals = [
        transaction["total"] for transaction in customer_transactions
        if transaction["total"] is not None
    ]

    print(
        customer,
        "| Transactions:", len(customer_transactions),
        "| Valid transactions:", len(valid_totals),
        "| Total spending:", sum(valid_totals)
    )


# 6. Product summaries
print("\n6. PRODUCT SUMMARIES")

for product in product_names:
    product_transactions = [
        transaction for transaction in transactions
        if transaction["product"] == product
    ]

    valid_totals = [
        transaction["total"] for transaction in product_transactions
        if transaction["total"] is not None
    ]

    valid_quantities = [
        transaction["quantity"] for transaction in product_transactions
        if transaction["quantity"] is not None
    ]

    print(
        product,
        "| Transactions:", len(product_transactions),
        "| Units sold:", sum(valid_quantities),
        "| Total sales:", sum(valid_totals)
    )


# 7. Ordered transaction results
print("\n7. ORDERED TRANSACTION RESULTS")

ordered_transactions = sorted(
    transactions,
    key=lambda transaction: (
        transaction["total"] is None,
        -(transaction["total"] or 0),
        transaction["transaction_id"]
    )
)

for transaction in ordered_transactions:
    print(
        transaction["transaction_id"],
        "| Customer:", transaction["customer"],
        "| Product:", transaction["product"],
        "| Quantity:", transaction["quantity"],
        "| Total:", transaction["total"],
        "| Category:", get_transaction_category(transaction["total"])
    )


# 8. Final summary
print("\n8. FINAL SUMMARY")

valid_totals = [
    transaction["total"] for transaction in transactions
    if transaction["total"] is not None
]

print("Total records:", len(transactions))
print("Unique records:", len(seen))
print("Repeated transactions:", len(repeated_transactions))
print("Incomplete transactions:", len(incomplete_transactions))
print("Total sales:", sum(valid_totals))
print("Highest transaction:", max(valid_totals) if valid_totals else "N/A")
print("Lowest transaction:", min(valid_totals) if valid_totals else "N/A")
print("Transaction categories:", category_counts)
```
